### Implementation of Basic Solution

```python
# for each submissions, create a set of words
def create_word_sets(submissions):
    word_sets = {}
    for key, submission in submissions.items():
        word_sets[key] = set(split_into_words(submission))
    return word_sets
```

```python
# calculate jaccard similarity for each pair of submissions
def calculate_jaccard_similarities(word_sets):
    similarities = {}
    keys = list(word_sets.keys())
    for i in range(len(keys)):
        for j in range(i + 1, len(keys)):
            key1 = keys[i]
            key2 = keys[j]
            sim = jaccard_similarity(word_sets[key1], word_sets[key2])
            similarities[(key1, key2)] = sim
    return similarities
```

```python
# utility function: print actual contents too and highlight in green the common words
def highlight_common_words(submission1, submission2):
    words1 = set(submission1.split())
    words2 = set(submission2.split())
    common_words = words1.intersection(words2)
    
    highlighted1 = ' '.join([f"\033[92m{word}\033[0m" if word in common_words else word for word in submission1.split()])
    highlighted2 = ' '.join([f"\033[92m{word}\033[0m" if word in common_words else word for word in submission2.split()])
    
    return highlighted1, highlighted2
    
```

```python
# generate k-shingles from list of words
def generate_k_shingles(words, k):
    """
    Generate k-shingles from a list of words.
    A k-shingle is a contiguous sequence of k words.
    """
    shingles = set()
    for i in range(len(words) - k + 1):
        shingle = ' '.join(words[i:i + k])
        shingles.add(shingle)
    return shingles 
```

```python
# implement hash function using ax+b mod p
def hash_function(word, a, b, p):
    """
    Hash function for a word using the formula (a * hash(word) + b) % p
    where p is a prime number.
    """
    return (a * hash(word) + b) % p
```

```python
hash_function("apple", 3, 5, 11)  # Example usage of the hash function
```

```python
# obtain n number of hash functions 
def generate_hash_functions(n, p):
    """
    Generate n hash functions using the formula (a * hash(word) + b) % p
    where p is a prime number.
    """
    import random
    hash_functions = []
    for _ in range(n):
        a = random.randint(1, p - 1)
        b = random.randint(0, p - 1)
        hash_functions.append((a, b))
    return hash_functions
```

```python
def create_vocabulary_based_on_k_shingles(k, submissions):
    """
    Create a vocabulary based on k-shingles from the sample submissions.
    """
    vocabulary = set()
    for submission in submissions.values():
        words = split_into_words(submission)
        shingles = generate_k_shingles(words, k)
        vocabulary.update(shingles)
    return sorted(vocabulary)
```

```python
# create a vector based representation of submissions based on vocabulary
def create_vector_representation(submissions, k_shingles):
    """
    Create a vector representation of submissions based on the vocabulary.
    Each submission is represented as a vector of word frequencies.
    """
    if k_shingles != 0:
        vocabulary = create_vocabulary_based_on_k_shingles(k_shingles, submissions)
    else:
        vocabulary = create_vocabulary(submissions)
    vectors = {}
    for key, submission in submissions.items():
        if k_shingles != 0:
            words = list(generate_k_shingles(split_into_words(submission), k_shingles))
        else:
            words = split_into_words(submission)
        vector = [words.count(word) for word in vocabulary]
        vectors[key] = vector
    return vectors
```

```python
# implement minhashing
def minhashing(shingles, hash_function_parameters, p):
    """
    Perform MinHashing on a set of shingles.
    :param shingles: Set of shingles
    :param num_hashes: Number of hash functions to use
    :param p: A prime number for the modulus operation
    :return: List of minhash values
    """
    minhashes = [float('inf')] * len(hash_function_parameters)
    for shingle in shingles:
        for i in range(len(hash_function_parameters)):
            a, b = hash_function_parameters[i]
            hash_value = hash_function(shingle, a, b, p)
            if hash_value < minhashes[i]:
                minhashes[i] = hash_value
    return minhashes
```
